```python
# Leitura do CSV cercado por um tratador de exceções
try:
    votaçao_por_municipio_df = pd.read_csv('votacao_candidato.csv', sep= ';', encoding='iso-8859-1')
    print('CSV lido com sucesso')
except FileNotFoundError:
    print('Arquivo não encontrado')
```

```python
# Validação de Granularidade: Antes de realizar o merge, foi isolado o município de BUJARI para checar a presença de múltiplas 
# linhas por conta das zonas eleitorais e validar os tipos das chaves.

# teste = votaçao_por_municipio_df[votaçao_por_municipio_df['Município'] == 'BUJARI']
# teste.head()
```

```python
# votaçao_por_municipio_df.info()
```

```python
votaçao_por_municipio_df['Data de carga'] = pd.to_datetime(votaçao_por_municipio_df['Data de carga'])
```

```python
# votaçao_por_municipio_df.info()
```

```python
try:
    cod_municipio = pd.read_csv('municipio_tse_ibge.csv', sep= ';', encoding='iso-8859-1')
    print('CSV lido com sucesso')
except FileNotFoundError:
    print('Arquivo não encontrado')
```

```python
cod_municipio.head()
```

```python
# cod_municipio.info()
```

```python
votacao_municipio_cod_df = votaçao_por_municipio_df.merge(cod_municipio, left_on= 'Código município', right_on='CD_MUNICIPIO_TSE')
```

```python
votacao_municipio_cod_df.head()
```

```python
# votacao_municipio_cod_df.info()
```

```python
vot_mun_cod_df_agrupado = votacao_municipio_cod_df.groupby(['CD_MUNICIPIO_IBGE','Município','UF', 'Turno', 'Nome candidato', 'Partido', 'Cargo']).agg({'Votos válidos' : 'sum', 'Votos nominais' : 'sum'}).reset_index().sort_values(by='Votos nominais', ascending=False)
```

```python
vot_mun_cod_df_agrupado.head()
```

```python
vot_mun_cod_df_agrupado.info()
```

```python
try:
    municipios_df = pd.read_csv('dim_municipio_socioeconomico.csv', sep= ',', encoding='iso-8859-1')
    print('CSV lido com sucesso')
except FileNotFoundError:
    print('Arquivo não encontrado')
```

```python
municipios_df.head()
```

```python
df_fato_final = vot_mun_cod_df_agrupado.merge(municipios_df, on= 'CD_MUNICIPIO_IBGE', how= 'left').reset_index()
```

```python
df_fato_final.head()
```

```python
df_fato_final.info()
```

```python
# Limpeza de nulos
df_fato_final = df_fato_final.dropna(subset=['idhm'])
df_fato_final.info()
```

```python
df_fato_final.to_csv('fato_votacao_socioeconomica_2022.csv', sep=';', index=False, encoding='utf-8')
print("Tabela Fato gerada com sucesso! Exportada e pronta pro Power BI.")
```
